## Step 4 - Try your own text

```
from predictor import FakeNewsPredictor
p = FakeNewsPredictor('model')
for t in [
    'Narendra Modi is the Prime Minister of India',
    'The government announced on Tuesday that a new education policy will provide financial assistance to students, officials said.',
    'SHOCKING!!! You will not believe what this celebrity said about the vaccine, the media is hiding it!',
]:
    r = p.predict(t)
    print(f"{r['label']:>9} {r['confidence']:.1%} | {t[:80]}")
```
